```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Buat data x sebanyak 500 titik dari -4 ke 4
x = np.linspace(-4, 4, 500)
y = norm.pdf(x)  # Hitung densitas distribusi normal

# Buat plot
plt.plot(x, y, label='Distribusi Normal', color='blue')
plt.axvline(x=0, color='green', linestyle='--', linewidth=2, label='Sumbu Tengah (x=0)')
plt.xlabel('x')
plt.ylabel('density')
plt.title('Kurva Distribusi Normal Standar')
plt.legend()
plt.grid(True)
plt.show()
```
